```python
import shlex

# Pon en None cualquier modelo DL que no quieras visualizar (basta con 1)
CTGCE_WEIGHTS = "/content/drive/MyDrive/CrowdSegmenter-CT-GCE/results/ctgce/naip_wellpad/ct_gce_final.pth"
TGCE_SSPS_WEIGHTS = "/content/drive/MyDrive/CrowdSegmenter-CT-GCE/results/tgce_ssps/naip_wellpad/tgce_ssps_final.pth"
CROWDSEG_WEIGHTS = "/content/drive/MyDrive/CrowdSegmenter-CT-GCE/results/crowdseg/naip_wellpad/crowdseg_final.pth"
SUPERVISED_WEIGHTS = "/content/drive/MyDrive/CrowdSegmenter-CT-GCE/results/supervised/naip_wellpad/supervised_final.pth"

INCLUDE_MV = True
INCLUDE_STAPLE = True
METHODS = None
SPLIT = "test"
NUM_SAMPLES = 6
SAMPLE_INDICES = None  # Opcional, ej: [0, 12, 30]
SAVE_NPZ = True
OUTPUT_DIR = "/content/drive/MyDrive/CrowdSegmenter-CT-GCE/results/visualizations/naip_wellpad"

extra_flags = []
for flag, val in (
    ("--ctgce_weights", CTGCE_WEIGHTS),
    ("--tgce_ssps_weights", TGCE_SSPS_WEIGHTS),
    ("--crowdseg_weights", CROWDSEG_WEIGHTS),
    ("--supervised_weights", SUPERVISED_WEIGHTS),
):
    if val:
        extra_flags.extend([flag, shlex.quote(val)])
if not INCLUDE_MV:
    extra_flags.append("--no-mv")
if not INCLUDE_STAPLE:
    extra_flags.append("--no-staple")
if METHODS:
    extra_flags.append("--methods")
    extra_flags.extend(shlex.quote(m) for m in METHODS)
if SAMPLE_INDICES is not None:
    extra_flags.append("--sample_indices")
    extra_flags.extend(str(i) for i in SAMPLE_INDICES)
if SAVE_NPZ:
    extra_flags.append("--save-npz")
extra_cli = " ".join(extra_flags)

!python scripts/visualize_predictions.py --dataset naip_wellpad --data "{path}" --split {SPLIT} --num_samples {NUM_SAMPLES} --output_dir "{OUTPUT_DIR}" {extra_cli}
```
